```python
import matplotlib.pyplot as plt
import numpy as np

def plot_raw_sensor_data(data, feature_names=None, title="Raw Sensor Data"):
    """
    Quickly visualizes raw sensor data over time.

    Parameters
    ----------
    data          : array-like, shape (n_samples, n_channels)
    feature_names : list of str, optional (names for the legend)
    title         : str, title of the plot
    """
    # Set a nice, wide figure size for time-series data
    plt.figure(figsize=(12, 4))
    
    # Plot all columns at once
    lines = plt.plot(data)
    
    # Formatting
    plt.title(title, fontsize=14)
    plt.xlabel("Samples (Time)", fontsize=12)
    plt.ylabel("Sensor Value", fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.7)
    
    # Add a legend if channel names are provided
    if feature_names is not None:
        # Ensure we only apply names up to the number of columns we actually have
        plt.legend(lines, feature_names[:data.shape[1]], loc='upper right')
        
    plt.tight_layout()
    plt.show()
```
